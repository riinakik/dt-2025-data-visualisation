```python
import pandas as pd

df = pd.read_csv("Competencies.csv")
df 
```

```python
import matplotlib.pyplot as plt

avg_effort =df.groupby("Student")["Hours_Studied"].mean()
avg_effort.plot(kind="bar")

plt.title("Average effort by student")
plt.ylabel("Average hours studied")

plt.show()

```

```python
avg_hours = df.groupby("Student", as_index=False)["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)

plt.title(("Average effort by student"))
plt.ylabel("Average hours studied")

plt.xlabel("Name")

```

```python
avg_grade =df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
avg_grade.plot(kind="bar")

plt.title("Average grade by student")
plt.ylabel("Average garde")

plt.show()
```

```python
avg_grade =df.groupby("Student")["Grade"].mean().sort_values(ascending=False)

plt.figure(figsize=(8,8)) # For changing the size

#plot the pie chart

plt.pie(avg_grade, labels=avg_grade.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
#colors=plt.cm.Paired.colors Different colors for different slices
#startangle=140 Rotates the chart
#autopct='%1.1f%%' Percentages

plt.show()
```

More studing= better grades?

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*10, alpha=0.6)
plt.xlabel("Hours")
plt.ylabel("Grade")
plt.title("Grade vs hours studied")
plt.show()
```

```python
avg_course =df.groupby("Course")["Grade"].mean()

avg_course.plot(kind="bar", figsize=(8,5), color="orchid")

plt.title("Average grade by Course")
plt.ylabel("Grade")

plt.show()
```

```python
avg_program =df.groupby("Program")["Grade"].mean().sort_values(ascending=False)
avg_program.plot(kind="bar",figsize=(8,5), color="gold")

plt.title("Average grade by Program")
plt.ylabel("Grade")

plt.show()
```

How consistent are study patterns and performance?

```python
#calculate quartiles and extremes
#Grades
grade_min = df["Grade"].min()
grade_q1 = df["Grade"].quantile(0.25)
grade_q3 = df["Grade"].quantile(0.75)
grade_max = df["Grade"].max()

#Years of practice
hours_min = df["Hours_Studied"].min()
hours_q1 = df["Hours_Studied"].quantile(0.25)
hours_q3 = df["Hours_Studied"].quantile(0.75)
hours_max = df["Hours_Studied"].max()

print("Grade - Min:", grade_min, "Q1:", grade_q1, "Q3:", grade_q3, "Max:", grade_max)
print("Hours of practice - Min:", hours_min, "Q1:", hours_q1, "Q3:", hours_q3, "Max:", hours_max)

plt.figure(figsize=(8,5))
df[["Grade", "Hours_Studied"]].boxplot()
plt.title("Box plot for grade and hours studied")
plt.show
```

```python
plt.figure(figsize=(8,5))
df[["Hours_Studied", "Grade"]].boxplot()
plt.title("Box plot to find how consistent are study patterns and performance")
plt.show
```
